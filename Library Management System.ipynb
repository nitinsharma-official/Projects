## Python Projects

    1. Library Management System
    
    A system to manage books, members, and issue/return operations.
    
    Features: Add, remove, update books; track borrowed books; manage member details.

```python
class Library:
    def __init__(self):
        self.data = {
            "books": [],
            "members": [],
            "borrowed": []
        }

    def add_book(self):
        book_id = int(input("Enter Book ID No.... "))
        title = input("Enter Book Title.... ")
        self.data["books"].append({"id": book_id, "title": title})
        print(f"Book {title} added successfully.... \n")

    def remove_book(self):
        book_id = int(input("Enter Book ID No.  to remove.... "))
        for book in self.data["books"]:
            if book["id"] == book_id:
                self.data["books"].remove(book)
                print(f"Book {book['title']} removed....\n")
                return
        print("Book not found....\n")

    def update_book(self):
        book_id = int(input("Enter Book ID No.  to update.... "))
        for book in self.data["books"]:
            if book["id"] == book_id:
                book["title"] = input("Enter new title.... ")
                print(f"Book updated successfully....\n")
                return
        print("Book not found....\n")

    def add_member(self):
        member_id = int(input("Enter Member ID No.... "))
        name = input("Enter Member Name.... ")
        self.data["members"].append({"id": member_id, "name": name})
        print(f"Member {name} added successfully....\n")

    def issue_book(self):
        book_id = int(input("Enter Book ID No. to issue.... "))
        member_id = input("Enter Member ID No.... ")

        for book in self.data["books"]:
            if book["id"] == book_id:
                self.data["borrowed"].append({"book_id": book_id, "member_id": member_id})
                self.data["books"].remove(book)
                print(f"Book {book['title']} issued to Member ID {member_id}...\n")
                return
        print(f"Book not available or already issued....\n")

    def return_book(self):
        book_id = int(input("Enter Book ID No. to return.... "))
        for record in self.data["borrowed"]:
            if record["book_id"] == book_id:
                self.data["borrowed"].remove(record)
                title = input("Enter Book Title.... ")
                self.data["books"].append({"id": book_id, "title": title})
                print(f"Book {title} returned successfully....\n")
                return
        print(f"Record not found....\n")

    def show_books(self):
        if not self.data["books"]:
            print(f"No books available....\n")
            return
        print(f"\nAvailable Books....")
        for book in self.data["books"]:
            print(f"ID:- {book['id']} | Title:- {book['title']}")
        print()

    def show_borrowed(self):
        if not self.data["borrowed"]:
            print(f"No borrowed books....\n")
            return
        print(f"\nBorrowed Books.... ")
        for record in self.data["borrowed"]:
            print(f"Book ID:- {record['book_id']} | Borrowed by Member ID:- {record['member_id']}")
        print()

library = Library()

while True:
    print("* JAI SHREE RAM *       \n   * Welcome *\n* Library Management System *")
    print("1. Add Book\n2. Remove\n3. Update\n4. Add Member\n5. Issue Book\n6. Return Book\n7. Show Available Book's\n8. Show Borrowed\n9. Exit")
    
    choice = input("Enter your choice 1 to 9.... ")
    print()

    if choice == "1":
        library.add_book()
    elif choice == "2":
        library.remove_book()
    elif choice == "3":
        library.update_book()
    elif choice == "4":
        library.add_member()
    elif choice == "5":
        library.issue_book()
    elif choice == "6":
        library.return_book()
    elif choice == "7":
        library.show_books()
    elif choice == "8":
        library.show_borrowed()
    elif choice == "9":
        print("Exit")
        break
    else:
        print(f"Invalid choice Please try again....\n")
```
