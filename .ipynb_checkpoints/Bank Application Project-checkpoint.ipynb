```python
data = {}
key = 1004
```

```python
import string as st
from getpass import getpass
def valid_password(password):
    if len(password) > 8:
        upper = 0
        special_charchter = 0
        digit = 0
        space = 0
        i = 0
        while i < len(password):
            if password[i].isupper():
                upper += 1

            elif password[i].isdigit():
                digit += 1
                
            elif password[i].isspace():
                space += 1

            elif password[i]in st.punctuation:
                special_charchter += 1


            i += 1
        if(upper > 0 and digit > 0 and special_charchter > 0) and (space == 0):
            return True
        else:
            return False
    else:
        False
```

```python
def sign_up():
    global key
    print("welcome To The SBI services :-")
    username = input("Enter your full name")
    password = input("Enter your strong password")
    confirm_password = input("Enter a confirm your password")

    if password == confirm_password:
        choice = input("Minimum Balance Required is 500 Press YES/NO").upper()
        if choice == "YES":
            data[key] = {
                "Name" : username,
                "Password" : password,
                "Balance" : 500
            }
            print(f"Your Account No. is {key}")
            key += 1
            print("Your Account Created Successfully")

        else:
            print("Minimum Balance Required")
            print("Thank you for vigiting SBI")

    else:
        print("Password do not match with previous Password")
```

```python
def log_in():
    print("Welcome To The SBI sevices")
    print("Login")
    AC_no = int(input("Enter Your Account Number"))
    if AC_no in data:
        password = input("Enter Your Password")
        if password == data[AC_no]["Password"]:
            print("Login Successfull")
        while True:
            choice = int(input("1.Credit\n2.Debit\n3.Change Password\n4.Show details\n5.Check Balance\n6.Logout\n"))
            if choice == 1:
                amount = int(input("Enter The amount you want to add..."))
                data[AC_no]["Balance"] += amount
                print("Amount Crebited...")

            elif choice == 2:
                amount = int(input("Enter the amount you want to debit..."))
                data[AC_no]["Balance"] -= amount
                print("Amount Debited...")
            elif choice == 3:                
                print("Changed Your Password")
                old_password = input("Enter Your Current Password... :- ")

                if old_password == data[AC_no]["Password"]:
                    new_password = input("Enter Your New Password... :-")
                    confirm_pass = input("Enter a confirm your Password... :-")

                    if new_password == confirm_pass:
                        if new_password:
                            data[AC_no]["Password"] = new_password
                            print("Your Password Changed Successfully...!")
                        else:
                            print("Incorrect Password...")
                    else:
                        print("New Password does not match with previous")
                else:
                    print("Incorrect current Password")

            elif choice == 4:
                print(f"A/C Holder Name {data[AC_no]["Name"]}")
            elif choice == 5:
                print(f"Your Current balance is {data[AC_no]["Balance"]}")

            elif choice == 6:
                print(f"Logout Successfully...")
                print("Thank you for visiting SBI")
                break

        else:
            print("Incorrect Password")
    else:
        print("Account does not exit")
```
